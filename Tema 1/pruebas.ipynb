```python
import requests

url = 'https://pokeapi.co/api/v2/pokemon/25'
respuesta = requests.get(url)
datos = respuesta.json()

# Mostramos solo algunas claves para no saturar la salida
print('Claves del JSON:', list(datos.keys()))
print('Nombre:',  datos['name'])
print('Peso (hectogramos):', datos['weight'])
print('Altura (decimetros):', datos['height'])
print('Tipos:', [t['type']['name'] for t in datos['types']])
print('Habilidades:', [h['ability']['name'] for h in datos['abilities']])
print('Estadisticas:')
for stat in datos['stats']:
    print(f"  {stat['stat']['name']}: {stat['base_stat']}")
```
