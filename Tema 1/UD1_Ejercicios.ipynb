## Ejercicio 1.2 — Pokémon y la PokéAPI

La [PokéAPI](https://pokeapi.co) es una API pública y gratuita con información de todos los Pokémon. Es un ejemplo perfecto de fuente de datos **externa y semiestructurada** que se consume mediante peticiones HTTP.

**Parte A — Exploración:**  
El código de abajo descarga la ficha del Pokémon número 25 (Pikachu). Ejecútalo y responde las preguntas que hay debajo.


````python
import requests

url = 'https://pokeapi.co/api/v2/pokemon/25'
respuesta = requests.get(url)
datos = respuesta.json()

# Mostramos solo algunas claves para no saturar la salida
print('Claves del JSON:', list(datos.keys()))
print('Nombre:',  datos['name'])
print('Peso (hectogramos):', datos['weight'])
print('Altura (decimetros):', datos['height'])
print('Tipos:', [t['type']['name'] for t in datos['types']])
print('Habilidades:', [h['ability']['name'] for h in datos['abilities']])
print('Estadisticas:')
for stat in datos['stats']:
    print(f"  {stat['stat']['name']}: {stat['base_stat']}")


````

**Preguntas sobre la Parte A:**

1. ¿Por qué decimos que este JSON es datos *semiestructurados* y no estructurados?
2. ¿Qué deberías hacer para convertir estos datos en un DataFrame de pandas con una fila por Pokémon y columnas como `nombre`, `peso`, `altura`, `tipo_1`, `hp`, `ataque`?

*Responde aquí:*

1. Decimos que el JSON es semiestructurado porque la información está organizada mediante claves y valores, por ejemplo name, weight, height o types. No está organizada directamente en filas y columnas como una tabla, pero sí sigue una estructura.

2. Para convertir estos datos en un DataFrame de pandas, primero habría que extraer del JSON solo los datos que nos interesan, como el nombre, peso, altura, tipo, vida y ataque. Después, se guardarían en un diccionario con campos como nombre, peso, altura, tipo_1, hp y ataque. Finalmente, se añadiría ese diccionario a una lista y se convertiría la lista en un DataFrame usando pandas. Así, cada Pokémon sería una fila y cada dato una columna.



**Parte B — A programar:**  
Descarga los datos de **5 Pokémon** de tu elección (puedes usar su número o su nombre) y construye un DataFrame de pandas con al menos las columnas: `nombre`, `peso`, `altura`, `hp`, `ataque`, `defensa`, `tipo_1`.
Muestra el DataFrame al final.


````python
import pandas as pd
import requests

# TODO: elige tus 5 Pokémon (por número o nombre)
pokemon_ids = [1, 4, 7, 25, 133]  # ej: [1, 4, 7, 25, 133]

registros = []
for pid in pokemon_ids:
    respuesta = requests.get(f"https://pokeapi.co/api/v2/pokemon/{pid}")
    if respuesta.status_code == 200:
        data = respuesta.json()
        registro = {
            "id": data["id"],
            "name": data["name"],
            "height": data["height"],
            "weight": data["weight"]
        }
        registros.append(registro)

df_pokemon = pd.DataFrame(registros)
print(df_pokemon)
````

## Ejercicio 2.4 — XML del mundo real: datos sísmicos del IGN

El IGN publica los terremotos ocurridos en España durante los **últimos diez días** en un canal **GeoRSS**: un documento XML con formato RSS 2.0 al que se añaden etiquetas geográficas. Vamos a leer ese XML y a extraer información de los terremotos más recientes.

La URL del canal es:
```
https://www.ign.es/ign/RssTools/sismologia.xml
```

Cada terremoto es un elemento `<item>` dentro de `<channel>` con este aspecto:

```xml
<item>
  <title>-Info.terremoto: 02/10/2026 4:06:42</title>
  <link>http://www.ign.es/web/ign/portal/sis-catalogo-terremotos/...?evid=es2026thaub</link>
  <description>Se ha producido un terremoto de magnitud 3.2 en SW PÁJARA.IFV en la fecha 02/10/2026 4:06:42 en la siguiente localización: 28.2395,-14.2982</description>
  <guid>http://www.ign.es/web/ign/portal/sis-catalogo-terremotos/...?evid=es2026thaub</guid>
  <geo:lat>28.2395</geo:lat>
  <geo:long>-14.2982</geo:long>
</item>
```

Fíjate en dos cosas: las coordenadas van en etiquetas con **namespace** (`geo:`), y la magnitud, el lugar y la fecha **no tienen etiqueta propia**: vienen mezclados dentro del texto de `<description>`.

**Parte A:** descarga el XML y explora su estructura.

````python
import requests
import xml.etree.ElementTree as ET

url = 'https://www.ign.es/ign/RssTools/sismologia.xml'
respuesta = requests.get(url, timeout=15)
respuesta.raise_for_status()
respuesta.encoding = 'utf-8'   # el canal está en UTF-8 (tildes en los topónimos)
xml_texto = respuesta.text

# Guardamos el XML por si queremos inspeccionarlo
with open('sismos_ign.xml', 'w', encoding='utf-8') as f:
    f.write(xml_texto)

# Parseamos
root = ET.fromstring(respuesta.content)
print('Etiqueta raíz:', root.tag, '| versión RSS:', root.get('version'))

canal = root.find('channel')
print('Título del canal:', canal.findtext('title'))
print('Descripción:', canal.findtext('description'))

items = canal.findall('item')
print(f'\nNúmero de <item> (terremotos): {len(items)}')

print('\nEtiquetas hijas del primer <item>:')
for hijo in items[0]:
    print(f'  {hijo.tag:45} -> {hijo.text}')
````

**Parte B:** extrae los datos relevantes y conviértelos en un DataFrame.

Para cada terremoto queremos estas columnas: `evid` (identificador, al final del enlace), `fecha` (tipo fecha/hora), `magnitud` (número), `lugar` (texto), `latitud` y `longitud` (números).

- Las coordenadas salen de `<geo:lat>` y `<geo:long>`. Observa en la salida de la Parte A cómo escribe ElementTree esas etiquetas: con el namespace completo entre llaves. Usa el diccionario `ns` para buscarlas.
- La magnitud, el lugar y la fecha hay que **extraerlos del texto** de `<description>` con una expresión regular (módulo `re`).

````python
import re
import pandas as pd

# Namespace de las etiquetas geo:lat y geo:long (aparece en la raíz <rss>)
ns = {'geo': 'http://www.w3.org/2003/01/geo/wgs84_pos#'}

# Expresión regular para sacar los datos del texto de <description>
patron = re.compile(
    r'magnitud (?P<magnitud>[\d.]+) en (?P<lugar>.+?) en la fecha (?P<fecha>\d{2}/\d{2}/\d{4} \d{1,2}:\d{2}:\d{2})'
)

registros = []

for item in root.findall('.//item'):
    descripcion = item.findtext('description', default='')
    resultado = patron.search(descripcion)

    if resultado:
        enlace = item.findtext('link', default='')

        registros.append({
            'evid': enlace.split('evid=')[-1].split('&')[0],
            'fecha': resultado.group('fecha'),
            'magnitud': resultado.group('magnitud'),
            'lugar': resultado.group('lugar'),
            'latitud': item.findtext('geo:lat', namespaces=ns),
            'longitud': item.findtext('geo:long', namespaces=ns)
        })

df_sismos = pd.DataFrame(registros)

if not df_sismos.empty:
    for columna in ['magnitud', 'latitud', 'longitud']:
        df_sismos[columna] = df_sismos[columna].astype(float)

    df_sismos['fecha'] = pd.to_datetime(
        df_sismos['fecha'],
        format='%d/%m/%Y %H:%M:%S'
    )

print(f'Terremotos extraídos: {len(df_sismos)}')
print(df_sismos.dtypes)
df_sismos.head()
````
