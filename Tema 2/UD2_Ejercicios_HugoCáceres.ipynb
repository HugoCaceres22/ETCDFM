## Ejercicio 2.3 — De CSV a Excel con múltiples hojas

Retomamos el dataset de sismos del ejercicio anterior. Un geólogo te pide un fichero Excel con:

- **Hoja 1** (`Todos`): todos los sismos con las columnas `time`, `latitude`, `longitude`, `depth`, `mag`, `place`.
- **Hoja 2** (`Fuertes`): solo los de magnitud ≥ 6.0, ordenados de mayor a menor magnitud.
- **Hoja 3** (`Resumen`): tabla con el número de sismos y magnitud media agrupados por `magType`.

El Excel debe tener la primera fila congelada y la columna de magnitud resaltada en amarillo en la hoja `Fuertes`.


````python

import pandas as pd
from openpyxl.styles import PatternFill

# Asegúrate de tener df_sismos cargado del ejercicio anterior
# Si no, descárgalo de nuevo aquí

COLS = ['time', 'latitude', 'longitude', 'depth', 'mag', 'place']

# TODO: crea df_todos, df_fuertes y df_resumen
df_todos = df_sismos[COLS].copy()

# Excel no admite fechas con zona horaria
df_todos['time'] = pd.to_datetime(df_todos['time'], utc=True).dt.tz_localize(None)

df_fuertes = (
    df_todos[df_todos['mag'] >= 6.0]
    .sort_values('mag', ascending=False)
)

df_resumen = (
    df_sismos.groupby('magType', dropna=False)
    .agg(
        numero_sismos=('mag', 'size'),
        magnitud_media=('mag', 'mean')
    )
    .reset_index()
)

# Exportar a Excel con múltiples hojas
excel_path = 'sismos_informe.xlsx'

with pd.ExcelWriter(excel_path, engine='openpyxl') as writer:
    df_todos.to_excel(writer, sheet_name='Todos', index=False)
    df_fuertes.to_excel(writer, sheet_name='Fuertes', index=False)
    df_resumen.to_excel(writer, sheet_name='Resumen', index=False)

    amarillo = PatternFill(fill_type='solid', fgColor='FFFF00')

    for ws in writer.book.worksheets:
        ws.freeze_panes = 'A2'

    ws = writer.book['Fuertes']
    col_mag = [c.column for c in ws[1] if c.value == 'mag'][0]

    for row in ws.iter_rows(
        min_row=2,
        min_col=col_mag,
        max_col=col_mag
    ):
        row[0].fill = amarillo

print(f'Fichero guardado: {excel_path}')

# Verificación: lee el Excel de vuelta y muestra las hojas
import openpyxl
wb = openpyxl.load_workbook(excel_path)
print(f'Hojas: {wb.sheetnames}')
````

## Ejercicio 3.1 — Benchmark CSV vs formatos binarios

Vamos a generar un dataset realista y comparar empíricamente cuánto espacio y tiempo tarda cada formato. Así verás por qué Parquet domina los pipelines de ML.

**Parte A — Genera y guarda:**


````python
import pandas as pd
import numpy as np
import time, os

rng = np.random.default_rng(42)
N = 200_000  # 200 000 filas

categorias = ['Electronica', 'Ropa', 'Hogar', 'Deporte', 'Libros', 'Alimentacion']
ciudades   = ['Madrid', 'Barcelona', 'Valencia', 'Sevilla', 'Bilbao', 'Zaragoza']
regiones   = ['Centro', 'Nordeste', 'Este', 'Sur', 'Norte', 'Aragón']

df = pd.DataFrame({
    'id_pedido':  np.arange(1, N+1),
    'fecha':      pd.date_range('2022-01-01', periods=N, freq='10min'),
    'cliente_id': rng.integers(1, 50000, N),
    'categoria':  rng.choice(categorias, N),
    'ciudad':     rng.choice(ciudades, N),
    'region':     rng.choice(regiones, N),
    'unidades':   rng.integers(1, 20, N),
    'precio_u':   rng.uniform(5, 500, N).round(2),
    'descuento':  rng.uniform(0, 0.4, N).round(3),
    'total':      None,  # calculado abajo
})
df['total'] = (df['precio_u'] * df['unidades'] * (1 - df['descuento'])).round(2)

print(f'Dataset: {df.shape}  |  columnas: {list(df.columns)}')

# TODO: guarda el dataset en los siguientes formatos:
# CSV
df.to_csv('benchmark.csv', index=False)

# Parquet (snappy)
df.to_parquet('benchmark_snappy.parquet', compression='snappy', index=False)

# Parquet (gzip)
df.to_parquet('benchmark_gzip.parquet', compression='gzip', index=False)

# Parquet (zstd)
df.to_parquet('benchmark_zstd.parquet', compression='zstd', index=False)

# Feather
df.to_feather('benchmark.feather')

print('Ficheros guardados. Continúa con la Parte B.')
````

**Parte B — Benchmark de tamaño y velocidad de lectura:**

**Parte B — Parsea y extrae a DataFrame:**  
El XML del IGN usa el namespace `http://quakeml.org/xmlns/bed/1.2`. Extrae `tiempo`, `latitud`, `longitud`, `profundidad_km` y `magnitud` de cada terremoto y construye un DataFrame.


````python

import pandas as pd
import xml.etree.ElementTree as ET
import re
import os

# Namespace del XML del IGN
NS = {'geo': 'http://www.w3.org/2003/01/geo/wgs84_pos#'}

if not os.path.exists('sismos_ign.xml'):
    raise FileNotFoundError('Falta sismos_ign.xml: ejecuta antes la Parte A con conexión a internet')

root = ET.parse('sismos_ign.xml').getroot()

# TODO: recorre los elementos <item> y extrae los campos
registros = []

for evento in root.findall('.//item'):
    descripcion = evento.findtext('description', '')

    fecha = re.search(r'\d{2}/\d{2}/\d{4} \d{2}:\d{2}:\d{2}', descripcion)
    mag = re.search(r'magnitud (\d+(?:[.,]\d+)?)', descripcion)

    lat = evento.findtext('geo:lat', namespaces=NS)
    lon = evento.findtext('geo:long', namespaces=NS)

    registros.append({
        'tiempo': fecha.group(0) if fecha else None,
        'latitud': float(lat) if lat else None,
        'longitud': float(lon) if lon else None,
        'profundidad_km': None,
        'magnitud': float(mag.group(1).replace(',', '.')) if mag else None
    })

df_ign = pd.DataFrame(registros)
print(f'Extraídos: {len(df_ign)} sismos')
print(df_ign.head())
````

````python

import xml.etree.ElementTree as ET

arbol = ET.parse('../Bloque_4_XML/datos/catalogo_libros.xml')
raiz = arbol.getroot()
print('Libros antes:', len(raiz.findall('libro')))

# --- Parte A: modificar ---
for libro in raiz.findall('libro'):
    precio = libro.find('precio')
    stock = int(libro.findtext('stock'))
    # TODO 1: si el género es 'Tecnico', sube el precio un 5 % (precio.text es un str)
    if libro.findtext('genero') == 'Tecnico':
        precio.text = f"{float(precio.text) * 1.05:.2f}"

    # TODO 2: pon el atributo 'disponible' a 'false' o 'true' según el stock (libro.set(...))
    if stock == 0:
        libro.set('disponible', 'false')
    else:
        libro.set('disponible', 'true')

# --- Parte B: borrar ---
# Ojo: no borres elementos mientras recorres la misma lista: recorre primero y borra después
descatalogados = []  # TODO: añade aquí los libros con stock 0 y año < 1900
for libro in raiz.findall('libro'):
    stock = int(libro.findtext('stock'))
    ano = int(libro.findtext('ano'))
    if stock == 0 and ano < 1900:
        descatalogados.append(libro)

for libro in descatalogados:
    raiz.remove(libro)
print('Libros borrados:', len(descatalogados))
print('Libros después:', len(raiz.findall('libro')))

# --- Parte C: guardar ---
ET.indent(arbol, space='  ')
arbol.write('catalogo_actualizado.xml', encoding='utf-8', xml_declaration=True)
````

## Ejercicio 6.3 — Seudonimizar antes de compartir

El equipo de ciencia de datos quiere entrenar un modelo de segmentación de clientes con `usuarios_anidado.json`, pero **no debe ver datos personales**. Genera el fichero `usuarios_ml.csv` (y su versión `usuarios_ml.json`) con **una fila por usuario** y estas columnas:

| Columna | Cómo se obtiene |
|---|---|
| `id_seud` | Seudónimo del correo con **HMAC-SHA256** y una clave secreta (primeros 16 caracteres hexadecimales) |
| `rango_edad` | Edad generalizada en tramos: `18-24`, `25-34`, `35-44`, `45-54`, `55+` |
| `provincia_cp` | Solo los **dos primeros dígitos** del código postal |

La clave secreta **no se escribe en el cuaderno**: se lee de la variable de entorno `ETCDFM_CLAVE_SEUD`. En Windows se define una vez con `setx ETCDFM_CLAVE_SEUD "una-clave-larga"` en la terminal; después hay que cerrar y volver a abrir VS Code para que el cuaderno la vea. Si no existe, el código genera una clave aleatoria para esta sesión.

````python
import hashlib
import hmac
import os
import secrets
import json
import pandas as pd

CLAVE = os.environ.get('ETCDFM_CLAVE_SEUD')
if CLAVE is None:
    CLAVE = secrets.token_hex(32)
    print('Aviso: no hay ETCDFM_CLAVE_SEUD; se usa una clave aleatoria solo para esta sesión.')

def seudonimo(valor: str) -> str:
    """Devuelve un seudónimo estable de 'valor' usando HMAC-SHA256 con CLAVE."""
    # TODO: usa hmac.new(CLAVE.encode(), valor.encode(), hashlib.sha256).hexdigest()
    #       y devuelve los primeros 16 caracteres
    return hmac.new(CLAVE.encode(), valor.encode(), hashlib.sha256).hexdigest()[:16]

def rango_edad(edad: int) -> str:
    # TODO: devuelve '18-24', '25-34', '35-44', '45-54' o '55+'
    if edad <= 24:
        return '18-24'
    elif edad <= 34:
        return '25-34'
    elif edad <= 44:
        return '35-44'
    elif edad <= 54:
        return '45-54'
    else:
        return '55+'

with open('../Bloque_5_JSON/datos/usuarios_anidado.json', encoding='utf-8') as f:
    usuarios = json.load(f)

filas = []
for u in usuarios:
    compras = u.get('historial_compras', [])
    filas.append({
        'id_seud': seudonimo(u['email']),
        'rango_edad': rango_edad(u['edad']),
        # TODO: provincia_cp, ciudad, n_compras, gasto_total, categorias_favoritas
        'provincia_cp': str(u['direccion']['codigo_postal'])[:2],
        'ciudad': u['direccion']['ciudad'],
        'n_compras': len(compras),
        'gasto_total': sum(c['total_eur'] for c in compras),
        'categorias_favoritas': ';'.join(u['preferencias']['categorias_favoritas'])
    })

df_ml = pd.DataFrame(filas)
print(df_ml.head())

# TODO: guarda df_ml en 'usuarios_ml.csv' (sin índice) y en 'usuarios_ml.json' (orient='records')
df_ml.to_csv('usuarios_ml.csv', index=False)
df_ml.to_json('usuarios_ml.json', orient='records', force_ascii=False)

# Comprobación: en el fichero generado no puede quedar ningún correo ni nombre
texto = df_ml.to_csv(index=False)
fugas = [u['email'] for u in usuarios if u['email'] in texto] + \
        [u['nombre'] for u in usuarios if u['nombre'] in texto]
print('Datos personales encontrados:', fugas if fugas else 'ninguno')
````
